```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.api as sm
import statsmodels.formula.api as smf

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
url = "https://raw.githubusercontent.com/selva86/datasets/master/Advertising.csv"
df = pd.read_csv(url)

df.head()
```

```python
df.describe()
```

```python
sns.pairplot(df)
plt.show()
```

```python
X = df[['TV', 'radio', 'newspaper']]
y = df['sales']

X = sm.add_constant(X)  # IMPORTANT: statsmodels does NOT add intercept automatically

model = sm.OLS(y, X)
results = model.fit()

results.summary()
```

```python
model_formula = smf.ols(
    formula="sales ~ TV + radio + newspaper",
    data=df
)
results_formula = model_formula.fit()

results_formula.summary()
```

```python
results.params
```

```python
results.conf_int()
```

```python
residuals = results.resid
fitted = results.fittedvalues

plt.scatter(fitted, residuals)
plt.axhline(0, color='red')
plt.xlabel("Fitted values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted")
plt.show()
```

```python
sm.qqplot(residuals, line='45')
plt.title("QQ Plot of Residuals")
plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif = pd.DataFrame()
vif["feature"] = X.columns
vif["VIF"] = [variance_inflation_factor(X.values, i)
              for i in range(X.shape[1])]

vif
```

```python
# Test if TV and Radio coefficients are jointly zero
hypothesis = "TV = radio = 0"
results.f_test(hypothesis)
```

```python
df['HighSales'] = (df['sales'] > df['sales'].median()).astype(int)

logit_model = smf.logit(
    "HighSales ~ TV + radio + newspaper",
    data=df
)
logit_results = logit_model.fit()

logit_results.summary()
```

```python
np.exp(logit_results.params)  # Odds ratios
```

```python
macro = sm.datasets.macrodata.load_pandas().data
macro['year'] = macro['year'].astype(int)
macro['quarter'] = macro['quarter'].astype(int)
macro['date'] = pd.PeriodIndex(
    year=macro.year,
    quarter=macro.quarter
).to_timestamp()

macro.set_index('date', inplace=True)

macro['realgdp'].plot(figsize=(10,4))
plt.title("Real GDP Over Time")
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

arima_model = ARIMA(macro['realgdp'], order=(1,1,1))
arima_results = arima_model.fit()

arima_results.summary()
```

```python
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(
    df[['TV','radio','newspaper']],
    y,
    test_size=0.2,
    random_state=42
)

lr = LinearRegression()
lr.fit(X_train, y_train)

preds = lr.predict(X_test)

mean_squared_error(y_test, preds)
```

### Statsmodels Interview Summary

- Statsmodels is for **statistical modeling and inference**
- Explicit intercept handling
- Rich `.summary()` output
- Strong diagnostics & hypothesis testing
- Ideal for econometrics & explainability
- Often paired with sklearn in real projects
